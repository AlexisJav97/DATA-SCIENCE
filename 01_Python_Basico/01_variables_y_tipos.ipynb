## FUNCIONES DE DEF
Una función sirve para agrupar instrucciones bajo un nombre y poder reutilizarlas cuando quieras.

`````python
def saludar():
    print("Hola")

saludar()
`````

`````python
def calcular_potencia(voltaje, corriente):
    potencia = voltaje * corriente
    print(f"Potenia: {potencia} W")

def calcular_corriente(voltaje, resistencia):
    corriente = voltaje / resistencia
    return corriente

corriente = calcular_corriente(24,3)
calcular_potencia(24, corriente)
`````

`````python
def calcular_promedio(lista):
    temp_total = sum(lista)
    cantidad_medidas = len(lista)

    promedio = temp_total/cantidad_medidas

    return promedio

temperaturas = [22,24,26,28,30]
promedio =  calcular_promedio(temperaturas)

print(f"Promedio: {promedio: .2f} °C")
`````

## ALCANCE DE VARIABLE: LOCAL Y GLOBAL
Cuando creas una variable dentro de una función, normalmente esa variable es local. Solo existe mientras se ejecuta esa función.

`````python
def calcular_potencia(voltaje, corriente):
    potencia = voltaje * corriente
    return potencia

resultado = calcular_potencia(24, 3)

print(resultado)
`````

```
PROGRAMA PRINCIPAL
│
├── resultado
│
└── calcular_potencia()
        │
        ├── voltaje
        ├── corriente
        └── potencia
             ↑
          variables
           locales
```

`````python
def analizar_temperaturas(lista):
    temp_total = sum(lista)
    num_medidas = len(lista)
    temp_max = max(lista)

    promedio_temp = temp_total/num_medidas
    return promedio_temp, temp_max


temperaturas = [22, 25, 28, 31, 27]
promedio, maximo = analizar_temperaturas(temperaturas)
resultado = analizar_temperaturas(temperaturas)

print(f"Temperatura promedio: {promedio:.2f} °C")
print(f"Temperatura maxima: {maximo:.2f} °C")
print(type(resultado))
`````

`````python
try:
    numero = float(input("Nùmero: "))

except ValueError:
    print("Valor incorrecto.")

else:
    print(f"Nùmero Vàlido: {numero}")

finally:
    print("Proceso terminado")
`````

`````python
# EL PROGRAMA DEBE PEDIR AL USUARIO INGRESA LAS RPM DEL MOTOR EN EL CASO DE INGRESAR UN VALOR INCORRECTO DEBE SALIR UN MENSAJE "Error: debes ingresar un número entero"

try:
    rpm = int(input("RPM del motor: "))
    if rpm < 0 :
        print("Error: Las RPM no pueden ser negativas")
    else:
        print(f"RPM ingresado: {rpm}")
except ValueError:
    print("Error: debes ingresar un número entero")


    
`````


`````python
while True:
    try:
        rpm = int(input("RPM del motor: "))
        if rpm < 0 :
            print("Error: Las RPM no pueden ser negativas")
            
        else:
            print(f"RPM ingresado: {rpm}")
            break
    except ValueError:
        print("Error: debes ingresar un número entero")

`````

`````python
while True:
    try:
        rpm = int(input("RPM del motor: "))
    except ValueError:
        print("Error: debes ingresar un numero entero")
    else:
        if rpm < 0:
            print("Las RPM deben ser postivas")
        else:
            print(f"RPM INGRESADAS: {rpm}")
            break

    finally:
        print("Intento FINALIZADO")
`````
